```python
# ===================================================
# PROJECT 9: AI CYBER-SECURITY FRAUD & SPAM DETECTOR
# AUTHOR: MORTOJA SARDAR (DATA ANALYST)
# ===================================================

!pip install gTTS -q
import time, IPython.display as ipd
from gtts import gTTS

print("=" * 60)
print("     🛡️ AI CYBER-SECURITY FRAUD DETECTOR SYSTEM 🛡️     ")
print("=" * 60 + "\n")

incoming_message = "CONGRATULATIONS! You won a $1000 lottery! Click this secret link right now to claim your cash reward!"

print(f"📥 [INCOMING TEXT] : \"{incoming_message}\"")
print("📊 [SYSTEM STATUS]  : Scanning for cybersecurity threats and phishing patterns...")
time.sleep(2)

message_lower = incoming_message.lower()
if "won" in message_lower or "lottery" in message_lower or "secret link" in message_lower or "cash reward" in message_lower:
    security_status = "CRITICAL SECURITY THREAT DETECTED (Phishing Fraud/Spam) 🚨"
    voice_msg = "Warning Mortoja! Critical security threat detected. The incoming message contains high-risk phishing fraud patterns. Blocking communication now."
else:
    security_status = "SECURE MESSAGE VERIFIED (Safe) ✅"
    voice_msg = "Mortoja, the incoming message is fully secure and safe."

print("-" * 60)
print(f"🔍 ANALYSIS TARGET   :  DATA INTEGRITY & SECURITY")
print("-" * 60)
print(f"📈 SECURITY OUTCOME  :  {security_status}")
print("-" * 60 + "\n")

tts = gTTS(voice_msg, lang='en')
tts.save("security_fraud_report.mp3")
ipd.Audio("security_fraud_report.mp3")
```
